```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Plot settings
plt.style.use("ggplot")
sns.set_theme(style="whitegrid")

print("Libraries Imported Successfully!")
```

```python
df = pd.read_csv(
    "../data/raw/skill_builder_data.csv",
    encoding="latin1",
    low_memory=False
)

print("Dataset Loaded Successfully!")
print("Shape :", df.shape)
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="correct"
)

plt.title("Correct vs Incorrect Answers")
plt.xlabel("Correct")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["attempt_count"],
    bins=30,
    kde=True
)

plt.title("Attempt Count Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["hint_count"],
    bins=30,
    kde=True
)

plt.title("Hint Count Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["ms_first_response"],
    bins=40,
    kde=True
)

plt.title("Response Time Distribution")

plt.show()
```

```python
top_skills = df["skill_name"].value_counts().head(15)

plt.figure(figsize=(14,6))

sns.barplot(
    x=top_skills.index,
    y=top_skills.values
)

plt.xticks(rotation=90)

plt.xlabel("Skill")
plt.ylabel("Number of Questions")

plt.title("Top 15 Most Frequent Skills")

plt.show()
```

```python
accuracy = (
    df.groupby("skill_name")["correct"]
      .mean()
      .sort_values()
      .head(15)
)

plt.figure(figsize=(14,6))

sns.barplot(
    x=accuracy.index,
    y=accuracy.values
)

plt.xticks(rotation=90)

plt.ylabel("Average Accuracy")

plt.title("Lowest Performing Skills")

plt.show()
```

```python
hints = (
    df.groupby("skill_name")["hint_count"]
      .mean()
      .sort_values(ascending=False)
      .head(15)
)

plt.figure(figsize=(14,6))

sns.barplot(
    x=hints.index,
    y=hints.values
)

plt.xticks(rotation=90)

plt.ylabel("Average Hint Count")

plt.title("Skills Requiring Most Hints")

plt.show()
```

```python
numeric_columns = df.select_dtypes(include=np.number)

plt.figure(figsize=(16,12))

sns.heatmap(
    numeric_columns.corr(),
    cmap="coolwarm",
    annot=False
)

plt.title("Correlation Matrix")

plt.show()
```

```python
plt.figure(figsize=(14,8))

numeric_columns.boxplot(rot=90)

plt.title("Boxplot of Numerical Features")

plt.show()
```
